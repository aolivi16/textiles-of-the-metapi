```python
import requests
```

```python
root = 'https://collectionapi.metmuseum.org'
search_path = '/public/collection/v1/search'
query = '?q=textile & isPublicDomain'
r = requests.get(f'{root}{search_path}{query}')
```

```python
r
```

```python
MET_parsed= r.json()
```

```python
MET_parsed
```

```python
first_fifteen = MET_parsed['objectIDs'][:15]
```

```python
for i in first_fifteen:
    url = f'{root}/public/collection/v1/objects/{i}'
    r1 = requests.get(url)
    p1 = r1.json()
    print(p1.get('title'))
    try:
        title = p1['title']
        print(title)
    except:
        title = 'none'
```

```python
First_hundred = MET_parsed['objectIDs'][:100]
```

```python
titles = []
mediums = []
cultures = []
departments = []
regions = []
```

```python
for i in First_hundred:
    url = f'{root}/public/collection/v1/objects/{i}'
    r1 = requests.get(url)
    p1 = r1.json()
    
    if p1.get('title'):
        titles.append(p1['title'])
    else:
        title = 'none'
        titles.append(title)
    
    if p1.get('medium'):
        mediums.append(p1['medium'])
    else:
        medium = 'none'
        mediums.append(medium)

    if p1.get('culture'):
        cultures.append(p1['culture'])
    else:
        culture = 'none'
        cultures.append(culture)

    if p1.get('department'):
        departments.append(p1['department'])
    else:
        department = 'none'
        departments.append(department)

    if p1.get('region'):
        regions.append(p1['region'])
    else:
        region = 'none'
        regions.append(region)
```

```python
import  pandas as pd
df = pd.DataFrame({
    'title' : titles,
    'medium' : mediums,
    'culture' : cultures,
    'department' : departments,
    'region' : regions
})
    
```

```python
df
```

```python
df.to_csv('textile_1.csv')
```
